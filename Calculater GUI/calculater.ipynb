```python
a=int(input("Enter a Number"))
b=input("Enter a Operator")
c=int(input("Enter a Number"))
if b=='+':
    print(a+c)
elif b=='-':
    print(a-c)
elif b=='*':
    print(a*c)
elif b=='/':
    print(a/c)
elif b=='%':
    print(a%c)  

    
```

```python

import tkinter as tk

root = tk.Tk()
root.title("Calculator")
root.geometry("220x250")

# Frame for display
display_frame = tk.Frame(root)
display_frame.pack(fill="x", padx=10, pady=10)

# Frame for buttons
button_frame = tk.Frame(root)
button_frame.pack(padx=10, pady=10)

display = tk.Entry(
    display_frame,
    width=25,
    font=("Arial", 20),
    justify="right"
)
display.pack(fill="x")

# Button functions
def click_1():
    display.insert("end", "1")

def click_2():
    display.insert("end", "2")

def click_3():
    display.insert("end", "3")

def click_4():
    display.insert("end", "4")

def click_5():
    display.insert("end", "5")

def click_6():
    display.insert("end", "6")

def click_7():
    display.insert("end", "7")

def click_8():
    display.insert("end", "8")
def click_9():
    display.insert("end", "9")

def click_10():
    display.insert("end", "0")

def click_11():
    display.insert("end", ".")

def click_12():
    expression = display.get()

    try:
        result = eval(expression)
        display.delete(0, "end")
        display.insert("end", result)

    except __________:
        display.delete(0, "end")
        display.insert("end", "Error")
def click_13():
    display.insert("end", "+")

def click_14():
    display.insert("end", "-")

def click_15():
    display.insert("end", "*")

def click_16():
    display.insert("end", "/")

def click_17():
    text = display.get()
    if text:
        display.delete(len(text) - 1, "end")

def click_18():
    display.delete(0, "end")

def click_19():
    display.delete(0, "end")

def click_20():
    display.insert("end", "%")

def click_21():
    display.insert("end", "+/-")


# Create buttons
btn1 = tk.Button(button_frame, text="1", width=8, command=click_1)
btn2 = tk.Button(button_frame, text="2", width=8, command=click_2)
btn3 = tk.Button(button_frame, text="3", width=8, command=click_3)
btn4 = tk.Button(button_frame, text="4", width=8, command=click_4)
btn5 = tk.Button(button_frame, text="5", width=8, command=click_5)
btn6 = tk.Button(button_frame, text="6", width=8, command=click_6)
btn7 = tk.Button(button_frame, text="7", width=8, command=click_7)
btn8 = tk.Button(button_frame, text="8", width=8, command=click_8)
btn9 = tk.Button(button_frame, text="9", width=8, command=click_9)
btn10 = tk.Button(button_frame, text="0", width=8, command=click_10)
btn11 = tk.Button(button_frame, text=".", width=8, command=click_11)
btn12 = tk.Button(button_frame, text="=",  bg="red",
    fg="white",width=8, command=click_12)
btn13 = tk.Button(button_frame, text="+", width=8, command=click_13)
btn14 = tk.Button(button_frame, text="-", width=8, command=click_14)
btn15 = tk.Button(button_frame, text="*", width=8, command=click_15)
btn16 = tk.Button(button_frame, text="/", width=8, command=click_16)
btn17 = tk.Button(button_frame, text="←", width=8, command=click_17)
btn18 = tk.Button(button_frame, text="C", width=8, command=click_18)

btn20 = tk.Button(button_frame, text="%", width=8, command=click_20)
btn21 = tk.Button(button_frame, text="+/-", width=8, command=click_21)

# Arrange buttons using grid
buttons = [
    [btn12,btn20, btn18, btn17],
    [btn7, btn8, btn9, btn16],
    [btn4, btn5, btn6, btn15],
    [btn1, btn2, btn3, btn14],
    [btn21, btn10, btn11, btn13]
]

for r, row_buttons in enumerate(buttons):
    for c, button in enumerate(row_buttons):
        button.grid(row=r, column=c, padx=3, pady=3, sticky="nsew")

for c in range(4):
    button_frame.grid_columnconfigure(c, weight=1)

root.mainloop()
```
